```python
# EXERCÍCIO 01 - CLASSIFICAÇÃO DE DESEMPENHO

matricula = input("Digite a matrícula do aluno: ").strip()
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))

media = (nota1 + nota2 + nota3) / 3

# Classificação da média (Evitando 4 ifs independentes)
if media < 4.0:
    situacao = "Reprovado"
elif media <= 5.9:
    situacao = "Recuperação"
elif media <= 7.9:
    situacao = "Aprovado"
else:
    situacao = "Aprovado com excelente desempenho"

# Informação personalizada pelo último dígito da matrícula
ultimo_digito = int(matricula[-1])
perfil = "Perfil A" if ultimo_digito % 2 == 0 else "Perfil B"

print("\n--- RELATÓRIO DO ALUNO ---")
print(f"Matrícula: {matricula}")
print(f"Notas: {nota1:.1f}, {nota2:.1f}, {nota3:.1f}")
print(f"Média: {media:.2f}")
print(f"Situação: {situacao}")
print(f"Perfil do Aluno: {perfil}")
```

```python
# EXERCÍCIO 02 - VALIDAÇÃO DE MATRÍCULA

matricula = input("Digite a matrícula do aluno (8 dígitos): ").strip()

motivos_invalidez = []

if len(matricula) != 8:
    motivos_invalidez.append("não possui exatamente 8 dígitos")
if not matricula.isdigit():
    motivos_invalidez.append("contém caracteres não numéricos")
elif matricula.startswith('0'):
    motivos_invalidez.append("o primeiro dígito não pode ser zero")

if not motivos_invalidez:
    primeiro_digito = matricula[0]
    ultimo_digito = matricula[-1]
    soma_digitos = sum(int(digito) for digito in matricula)

    if soma_digitos < 20:
        categoria = "Categoria 1"
    elif soma_digitos <= 39:
        categoria = "Categoria 2"
    else:
        categoria = "Categoria 3"

    print("\n--- MATRÍCULA VÁLIDA ---")
    print(f"Matrícula: {matricula}")
    print(f"Primeiro dígito: {primeiro_digito}")
    print(f"Último dígito: {ultimo_digito}")
    print(f"Soma dos dígitos: {soma_digitos}")
    print(f"Classificação: {categoria}")
else:
    print("\n--- MATRÍCULA INVÁLIDA ---")
    print("Motivo(s): " + "; ".join(motivos_invalidez) + ".")
```

```python
# EXERCÍCIO 03 - SISTEMA DE VOTAÇÃO

votos = []

print("--- SISTEMA DE VOTAÇÃO ---")
print("1 - Candidato A | 2 - Candidato B | 3 - Candidato C | 9 - Encerrar")

while True:
    voto = int(input("Informe o seu voto: "))
    if voto == 9:
        break
    votos.append(voto)

qtd_total = len(votos)
votos_a = 0
votos_b = 0
votos_c = 0
votos_invalidos = 0

# Contagem percorrendo a lista
for v in votos:
    if v == 1:
        votos_a += 1
    elif v == 2:
        votos_b += 1
    elif v == 3:
        votos_c += 1
    else:
        votos_invalidos += 1

pct_a = (votos_a / qtd_total * 100) if qtd_total > 0 else 0
pct_b = (votos_b / qtd_total * 100) if qtd_total > 0 else 0
pct_c = (votos_c / qtd_total * 100) if qtd_total > 0 else 0

# Determinação do candidato mais votado
if votos_a > votos_b and votos_a > votos_c:
    mais_votado = "Candidato A"
elif votos_b > votos_a and votos_b > votos_c:
    mais_votado = "Candidato B"
elif votos_c > votos_a and votos_c > votos_b:
    mais_votado = "Candidato C"
else:
    mais_votado = "Empate ou nenhum voto válido registrado"

print("\n--- RESULTADO DA VOTAÇÃO ---")
print(f"Total de votos: {qtd_total}")
print(f"Candidato A: {votos_a} voto(s) ({pct_a:.1f}%)")
print(f"Candidato B: {votos_b} voto(s) ({pct_b:.1f}%)")
print(f"Candidato C: {votos_c} voto(s) ({pct_c:.1f}%)")
print(f"Votos inválidos: {votos_invalidos}")
print(f"Candidato mais votado: {mais_votado}")
```

```python
# EXERCÍCIO 04 - CADASTRO E CLASSIFICAÇÃO DE IDADES

idades = []

print("--- CADASTRO DE IDADES (Digite 0 para encerrar) ---")
while True:
    idade = int(input("Digite a idade do aluno: "))
    if idade == 0:
        break
    if idade < 0:
        print("Erro: Idades negativas não são permitidas. Tente novamente.")
        continue

    idades.append(idade)

    if idade <= 12:
        classificacao = "Criança"
    elif idade <= 17:
        classificacao = "Adolescente"
    elif idade <= 59:
        classificacao = "Adulto"
    else:
        classificacao = "Idoso"
    print(f"Cadastrado: {idade} anos -> {classificacao}")

if idades:
    total_alunos = len(idades)
    maior_idade = max(idades)
    menor_idade = min(idades)
    media_idades = sum(idades) / total_alunos
    menores = sum(1 for i in idades if i < 18)
    maiores_18 = sum(1 for i in idades if i >= 18)
    acima_50 = sum(1 for i in idades if i > 50)

    print("\n--- ESTATÍSTICAS DA TURMA ---")
    print(f"Quantidade total de alunos: {total_alunos}")
    print(f"Maior idade: {maior_idade}")
    print(f"Menor idade: {menor_idade}")
    print(f"Média das idades: {media_idades:.1f}")
    print(f"Menores de idade (< 18): {menores}")
    print(f"Alunos com 18 anos ou mais: {maiores_18}")
    print(f"Alunos com idade superior a 50 anos: {acima_50}")
else:
    print("\nNenhum aluno foi cadastrado.")
```

```python
# EXERCÍCIO 05 - LISTA DE COMPRAS INTELIGENTE

produtos = []
CAPACIDADE_MAXIMA = 12

while True:
    print("\n--- LISTA DE COMPRAS INTELIGENTE ---")
    print("1 - Adicionar produto")
    print("2 - Remover produto")
    print("3 - Consultar produto")
    print("4 - Alterar produto")
    print("5 - Listar produtos")
    print("0 - Finalizar")

    opcao = input("Escolha uma opção: ").strip()

    if opcao == "1":
        if len(produtos) >= CAPACIDADE_MAXIMA:
            print("Capacidade máxima atingida (12 produtos)!")
            continue
        nome = input("Nome do produto: ").strip().lower()
        if not nome:
            print("Erro: O nome do produto não pode ser vazio.")
        elif nome in produtos:
            print("Erro: Produto já cadastrado.")
        else:
            produtos.append(nome)
            print(f"'{nome}' adicionado com sucesso!")

    elif opcao == "2":
        nome = input("Nome do produto a remover: ").strip().lower()
        if nome in produtos:
            produtos.remove(nome)
            print(f"'{nome}' removido com sucesso!")
        else:
            print("Erro: Produto não encontrado na lista.")

    elif opcao == "3":
        nome = input("Nome do produto a consultar: ").strip().lower()
        if nome in produtos:
            print(f"O produto '{nome}' ESTÁ cadastrado.")
        else:
            print(f"O produto '{nome}' NÃO está cadastrado.")

    elif opcao == "4":
        atual = input("Nome do produto atual: ").strip().lower()
        if atual in produtos:
            novo = input("Novo nome do produto: ").strip().lower()
            if not novo:
                print("Erro: O novo nome não pode ser vazio.")
            elif novo in produtos:
                print("Erro: O novo produto já existe na lista.")
            else:
                idx = produtos.index(atual)
                produtos[idx] = novo
                print(f"Produto '{atual}' alterado para '{novo}' com sucesso!")
        else:
            print("Erro: Produto atual não encontrado.")

    elif opcao == "5":
        print("\n--- PRODUTOS NA LISTA ---")
        if not produtos:
            print("A lista está vazia.")
        else:
            for i, p in enumerate(produtos, 1):
                print(f"{i}. {p}")

    elif opcao == "0":
        cadastrados = len(produtos)
        disponiveis = CAPACIDADE_MAXIMA - cadastrados
        print("\n--- RESUMO FINAL ---")
        print(f"Produtos cadastrados: {cadastrados}")
        print(f"Capacidade máxima: {CAPACIDADE_MAXIMA}")
        print(f"Espaços disponíveis: {disponiveis}")
        break
    else:
        print("Opção inválida. Tente novamente.")
```

```python
# EXERCÍCIO 06 - CONTROLE DE ESTOQUE

produtos = ["Arroz", "Feijao", "Macarrao", "Cafe", "Acucar"]
quantidades = [15, 8, 20, 5, 12]

while True:
    print("\n--- CONTROLE DE ESTOQUE ---")
    print("1 - Listar produtos")
    print("2 - Consultar estoque")
    print("3 - Entrada de produtos")
    print("4 - Saída de produtos")
    print("5 - Produtos com estoque baixo")
    print("0 - Sair")

    opcao = input("Escolha uma opção: ").strip()

    if opcao == "1":
        print("\n--- ESTOQUE ATUAL ---")
        for p, q in zip(produtos, quantidades):
            print(f"Produto: {p:<10} | Quantidade: {q}")

    elif opcao == "2":
        nome = input("Nome do produto: ").strip().capitalize()
        if nome in produtos:
            idx = produtos.index(nome)
            print(f"Estoque de {produtos[idx]}: {quantidades[idx]} unidades.")
        else:
            print("Produto não localizado no estoque.")

    elif opcao == "3":
        nome = input("Nome do produto: ").strip().capitalize()
        if nome in produtos:
            qtd = int(input("Quantidade recebida: "))
            if qtd < 0:
                print("Erro: Quantidade não pode ser negativa.")
            else:
                idx = produtos.index(nome)
                quantidades[idx] += qtd
                print(f"Entrada registrada! Novo estoque de {nome}: {quantidades[idx]}")
        else:
            print("Produto não localizado.")

    elif opcao == "4":
        nome = input("Nome do produto: ").strip().capitalize()
        if nome in produtos:
            qtd = int(input("Quantidade retirada: "))
            if qtd < 0:
                print("Erro: Quantidade não pode ser negativa.")
            else:
                idx = produtos.index(nome)
                if qtd > quantidades[idx]:
                    print(f"Erro: Estoque insuficiente (Disponível: {quantidades[idx]}).")
                else:
                    quantidades[idx] -= qtd
                    print(f"Saída registrada! Novo estoque de {nome}: {quantidades[idx]}")
        else:
            print("Produto não localizado.")

    elif opcao == "5":
        print("\n--- PRODUTOS COM ESTOQUE BAIXO (< 10) ---")
        encontrou = False
        for p, q in zip(produtos, quantidades):
            if q < 10:
                print(f"Produto: {p:<10} | Quantidade: {q}")
                encontrou = True
        if not encontrou:
            print("Nenhum produto está com estoque baixo.")

    elif opcao == "0":
        print("Saindo do programa de estoque...")
        break
    else:
        print("Opção inválida.")
```

```python
# EXERCÍCIO 07 - ANÁLISE DE DESEMPENHO DA TURMA

matriculas = []
acertos = []

print("--- DESEMPENHO DA TURMA (Digite matrícula 0 para encerrar) ---")
while True:
    mat = input("Matrícula do aluno: ").strip()
    if mat == "0":
        break
    try:
        ac = int(input("Quantidade de acertos (0 a 10): "))
        if 0 <= ac <= 10:
            matriculas.append(mat)
            acertos.append(ac)
        else:
            print("Erro: A quantidade de acertos deve ser entre 0 e 10.")
    except ValueError:
        print("Erro: Entrada inválida. Digite um número inteiro.")

if matriculas:
    total_alunos = len(matriculas)
    media_acertos = sum(acertos) / total_alunos
    maior_acertos = max(acertos)
    menor_acertos = min(acertos)

    aprovados = sum(1 for a in acertos if 8 <= a <= 10)
    recuperacao = sum(1 for a in acertos if 6 <= a <= 7)
    reprovados = sum(1 for a in acertos if 0 <= a <= 5)

    # Matrículas de quem teve a maior quantidade de acertos
    mats_maior = [matriculas[i] for i in range(total_alunos) if acertos[i] == maior_acertos]

    print("\n--- RESUMO DA TURMA ---")
    print(f"Média de acertos da turma: {media_acertos:.2f}")
    print(f"Maior quantidade de acertos: {maior_acertos}")
    print(f"Menor quantidade de acertos: {menor_acertos}")
    print(f"Alunos aprovados (8 a 10): {aprovados}")
    print(f"Alunos em recuperação (6 a 7): {recuperacao}")
    print(f"Alunos reprovados (0 a 5): {reprovados}")
    print(f"Matrícula(s) do(s) aluno(s) com maior nota: {', '.join(mats_maior)}")
else:
    print("\nNenhum aluno registrado.")
```

```python
# EXERCÍCIO 08 - PONTUAÇÃO PERSONALIZADA PELA MATRÍCULA

nome = input("Nome do aluno: ").strip()
matricula = input("Matrícula: ").strip()

notas = []
print("Informe as 5 notas do aluno:")
for i in range(1, 6):
    n = float(input(f"Nota {i}: "))
    notas.append(n)

media_original = sum(notas) / len(notas)

# Determinação do bônus pelo último dígito da matrícula
ultimo_digito = int(matricula[-1])
if 0 <= ultimo_digito <= 3:
    bonus = 0.5
elif 4 <= ultimo_digito <= 6:
    bonus = 1.0
else:
    bonus = 1.5

nota_final = min(10.0, media_original + bonus)

if nota_final < 5.0:
    situacao = "Reprovado"
elif nota_final < 7.0:
    situacao = "Recuperação"
elif nota_final <= 9.0:
    situacao = "Aprovado"
else:
    situacao = "Excelente"

maior_nota = max(notas)
menor_nota = min(notas)
acima_media = sum(1 for n in notas if n > media_original)
abaixo_media = sum(1 for n in notas if n < media_original)

print("\n--- RESULTADO FINAL ---")
print(f"Nome: {nome}")
print(f"Matrícula: {matricula}")
print(f"Notas: {notas}")
print(f"Média Original: {media_original:.2f}")
print(f"Bônus Aplicado: {bonus}")
print(f"Nota Final: {nota_final:.2f}")
print(f"Situação: {situacao}")
print(f"Maior nota individual: {maior_nota}")
print(f"Menor nota individual: {menor_nota}")
print(f"Notas acima da média original: {acima_media}")
print(f"Notas abaixo da média original: {abaixo_media}")
```

```python
# EXERCÍCIO 09 - SIMULAÇÃO DE CAIXA ELETRÔNICO

saldo = float(input("Informe o saldo inicial da conta R$: "))
extrato = []

while True:
    print("\n--- CAIXA ELETRÔNICO ---")
    print("1 - Consultar saldo")
    print("2 - Depositar")
    print("3 - Sacar")
    print("4 - Extrato")
    print("0 - Sair")

    opcao = input("Escolha uma opção: ").strip()

    if opcao == "1":
        print(f"Saldo atual: R$ {saldo:.2f}")

    elif opcao == "2":
        valor = float(input("Valor a depositar R$: "))
        if valor > 0:
            saldo += valor
            extrato.append(f"Deposito: R$ {valor:.2f}")
            print("Depósito realizado com sucesso!")
        else:
            print("Erro: O valor do depósito deve ser positivo.")

    elif opcao == "3":
        valor = float(input("Valor a sacar R$: "))
        if valor <= 0:
            print("Erro: O valor do saque deve ser positivo.")
        elif valor > saldo:
            print("Erro: Saldo insuficiente.")
        else:
            saldo -= valor
            extrato.append(f"Saque: R$ {valor:.2f}")
            print("Saque realizado com sucesso!")

    elif opcao == "4":
        print("\n--- EXTRATO ---")
        if not extrato:
            print("Nenhuma operação registrada.")
        else:
            for op in extrato:
                print(op)

        # Estatísticas percorrendo a lista extrato
        qtd_depositos = 0
        qtd_saques = 0
        total_depositado = 0.0
        total_sacado = 0.0

        for item in extrato:
            tipo, valor_str = item.split(": R$ ")
            val = float(valor_str)
            if tipo == "Deposito":
                qtd_depositos += 1
                total_depositado += val
            elif tipo == "Saque":
                qtd_saques += 1
                total_sacado += val

        print("\n--- RESUMO DO EXTRATO ---")
        print(f"Quantidade de depósitos: {qtd_depositos}")
        print(f"Quantidade de saques: {qtd_saques}")
        print(f"Total depositado: R$ {total_depositado:.2f}")
        print(f"Total sacado: R$ {total_sacado:.2f}")
        print(f"Saldo atual: R$ {saldo:.2f}")

    elif opcao == "0":
        print("Sessão finalizada com sucesso.")
        break
    else:
        print("Opção inválida.")
```

```python
# EXERCÍCIO 10 - SISTEMA DE ANÁLISE DE ALUNOS

matriculas = []
nomes = []
notas1 = []
notas2 = []
notas3 = []

def calcular_media(idx):
    return (notas1[idx] + notas2[idx] + notas3[idx]) / 3

def obter_situacao(media):
    if media >= 7.0:
        return "Aprovado"
    elif media >= 5.0:
        return "Recuperação"
    else:
        return "Reprovado"

while True:
    print("\n--- SISTEMA DE ANÁLISE DE ALUNOS ---")
    print("1 - Cadastrar aluno")
    print("2 - Listar alunos")
    print("3 - Consultar aluno")
    print("4 - Estatísticas da turma")
    print("5 - Alunos aprovados")
    print("6 - Alunos reprovados")
    print("7 - Aluno com maior média")
    print("0 - Encerrar")

    opcao = input("Escolha uma opção: ").strip()

    if opcao == "1":
        mat = input("Matrícula: ").strip()
        if mat in matriculas:
            print("Erro: Matrícula já cadastrada no sistema!")
            continue
        nome = input("Nome do aluno: ").strip()

        try:
            n1 = float(input("Nota 1 (0 a 10): "))
            n2 = float(input("Nota 2 (0 a 10): "))
            n3 = float(input("Nota 3 (0 a 10): "))
            if not (0 <= n1 <= 10 and 0 <= n2 <= 10 and 0 <= n3 <= 10):
                print("Erro: As notas devem estar entre 0 e 10.")
                continue
        except ValueError:
            print("Erro: Digite valores numéricos válidos para as notas.")
            continue

        matriculas.append(mat)
        nomes.append(nome)
        notas1.append(n1)
        notas2.append(n2)
        notas3.append(n3)
        print("Aluno cadastrado com sucesso!")

    elif opcao == "2":
        if not matriculas:
            print("Nenhum aluno cadastrado.")
        else:
            print("\n--- LISTA DE ALUNOS ---")
            for i in range(len(matriculas)):
                m = calcular_media(i)
                sit = obter_situacao(m)
                print(f"Matrícula: {matriculas[i]} | Nome: {nomes[i]} | Média: {m:.2f} | Situação: {sit}")

    elif opcao == "3":
        if not matriculas:
            print("Nenhum aluno cadastrado.")
            continue
        mat = input("Informe a matrícula para consulta: ").strip()
        if mat in matriculas:
            idx = matriculas.index(mat)
            m = calcular_media(idx)
            sit = obter_situacao(m)
            print(f"\nMatrícula: {matriculas[idx]}")
            print(f"Nome: {nomes[idx]}")
            print(f"Notas: {notas1[idx]:.1f}, {notas2[idx]:.1f}, {notas3[idx]:.1f}")
            print(f"Média: {m:.2f}")
            print(f"Situação: {sit}")
        else:
            print("Aluno não encontrado.")

    elif opcao == "4":
        if not matriculas:
            print("Nenhum aluno cadastrado.")
            continue
        medias = [calcular_media(i) for i in range(len(matriculas))]
        total = len(matriculas)
        media_geral = sum(medias) / total
        maior_m = max(medias)
        menor_m = min(medias)
        aprovados = sum(1 for m in medias if m >= 7.0)
        recuperacao = sum(1 for m in medias if 5.0 <= m < 7.0)
        reprovados = sum(1 for m in medias if m < 5.0)

        print("\n--- ESTATÍSTICAS DA TURMA ---")
        print(f"Quantidade de alunos: {total}")
        print(f"Média geral da turma: {media_geral:.2f}")
        print(f"Maior média: {maior_m:.2f}")
        print(f"Menor média: {menor_m:.2f}")
        print(f"Alunos Aprovados: {aprovados}")
        print(f"Alunos em Recuperação: {recuperacao}")
        print(f"Alunos Reprovados: {reprovados}")

    elif opcao == "5":
        if not matriculas:
            print("Nenhum aluno cadastrado.")
            continue
        print("\n--- ALUNOS APROVADOS (Média >= 7.0) ---")
        encontrou = False
        for i in range(len(matriculas)):
            m = calcular_media(i)
            if m >= 7.0:
                print(f"Matrícula: {matriculas[i]} | Nome: {nomes[i]} | Média: {m:.2f}")
                encontrou = True
        if not encontrou:
            print("Nenhum aluno aprovado.")

    elif opcao == "6":
        if not matriculas:
            print("Nenhum aluno cadastrado.")
            continue
        print("\n--- ALUNOS REPROVADOS (Média < 5.0) ---")
        encontrou = False
        for i in range(len(matriculas)):
            m = calcular_media(i)
            if m < 5.0:
                print(f"Matrícula: {matriculas[i]} | Nome: {nomes[i]} | Média: {m:.2f}")
                encontrou = True
        if not encontrou:
            print("Nenhum aluno reprovado.")

    elif opcao == "7":
        if not matriculas:
            print("Nenhum aluno cadastrado.")
            continue
        medias = [calcular_media(i) for i in range(len(matriculas))]
        maior_m = max(medias)
        print(f"\n--- ALUNO(S) COM MAIOR MÉDIA ({maior_m:.2f}) ---")
        for i in range(len(matriculas)):
            if medias[i] == maior_m:
                print(f"Matrícula: {matriculas[i]} | Nome: {nomes[i]}")
                print(f"Notas: {notas1[i]:.1f}, {notas2[i]:.1f}, {notas3[i]:.1f} | Média: {maior_m:.2f}\n")

    elif opcao == "0":
        print("Programa encerrado.")
        break
    else:
        print("Opção inválida. Tente novamente.")
```
